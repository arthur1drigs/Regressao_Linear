Atividade aluguel bicicletas


```python
from sklearn.linear_model import LinearRegression

duracao_aluguel = [[10],[15], [20],[25], [30], [35], [40], [50], [55], [60]]
preco_pago =   [10.0,  13.0, 14.0, 18.0, 19.0, 23.0, 24.0, 30.0, 32.0, 34.0 ]
	

model = LinearRegression()
model.fit(duracao_aluguel, preco_pago)


predicao = model.predict([[60]])

print(f"Para essa duração de tempo o custo previsto é de  {predicao} R$")


```

- Qual é a variável de entrada e qual é a variável que queremos prever?


A variável de entrada é a duração do aluguel e a variável que queremos prever é a preço pago.

- Qual preço o modelo estimou para 45 minutos?

O preço que o modelo estimou para 45 minutos é de 27,00 R$.

- As previsões para 20 e 60 minutos ficaram próximas dos valores históricos?


Sim a previsão de 20 e 60 minutos ficaram iguais os valores históricos.

- Porque o preço previsto deve ser apresentado como uma estimativa?


Por que o modelo fez as previsões com base nos dados dos históricos.

- Você teria a mesma confiança em uma previsão para 300 minutos? Justifique considerando os dados disponíveis.

Não porque o modelo só tem dados de 10 a 60 minutos.
